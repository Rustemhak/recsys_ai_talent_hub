```python
import pandas as pd
from implicit.nearest_neighbours import CosineRecommender
from rectools.dataset import Dataset
from rectools import Columns
from rectools.models.popular import PopularModel
from config.config_models import UserKnn_model_conf, Popular_model_conf
```

```python
interactions = pd.read_csv('../data/kion_train/interactions.csv')
users = pd.read_csv('../data/kion_train/users.csv')
items = pd.read_csv('../data/kion_train/items.csv')
```

```python
# rename columns, convert timestamp
interactions.rename(columns={'last_watch_dt': Columns.Datetime,
                            'total_dur': Columns.Weight},
                    inplace=True)

interactions['datetime'] = pd.to_datetime(interactions['datetime'])
```

```python
dataset = Dataset.construct(
            interactions_df=interactions,
            user_features_df=None,
            item_features_df=None,
        )
```

```python
pop = PopularModel()
pop.fit(dataset)
```

```python
popular_reco_df = pop.recommend(
    dataset.user_id_map.external_ids[:1],
    dataset=dataset,
    k=Popular_model_conf.N_recs,
    filter_viewed=False  # True - throw away some items for each user
)
```
